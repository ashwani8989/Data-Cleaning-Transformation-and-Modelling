Q1-  What is Data Loading in Power BI and why is it considered the first step of analysis?

Ans- Data loading in power BI is the prosess of inporting data from diffrent sources such a exl, my sql, csv file, websites, could servies for analysis and reporting.

Considered first step of analysis becouse-

1. Data must be available before analysis can begin.

2. It helps identify data quality issues early.

3. It connects business data to Power BI.

4. Accurate analysis depends on correct data loading.

Q2- Explain the difference between “Load” and “Transform Data” in Power BI.

Ans- Load- Importing an Excel sheet directly into Power BI without modifications.

   Transform Data- Removing null values, changing date formats, or splitting customer names before loading.


   | Feature         | Load                                | Transform Data                                       |
| --------------- | ----------------------------------- | ---------------------------------------------------- |
| Meaning         | Imports data directly into Power BI | Opens Power Query Editor to clean and modify data    |
| Purpose         | Quick data import                   | Data cleaning and preparation                        |
| Changes Allowed | No major changes before loading     | Allows filtering, splitting, merging, renaming, etc. |
| Tool Used       | Load option                         | Power Query Editor                                   |
| Time Taken      | Faster                              | May take more time due to processing                 |
| Best Used When  | Data is already clean               | Data contains errors or needs formatting             |

Q4- Why is Star Schema preferred over Snowflake Schema in Power BI?

Ans- Star schema preferred in power Bi becouse it provides better performence, easier understanding, and simpler relationships.

Reasons star schema preferred over snowflake schema.

| Star Schema                 | Snowflake Schema                        |
| --------------------------- | --------------------------------------- |
| Simple structure            | Complex structure                       |
| Faster query performance    | Slower due to multiple joins            |
| Easy to understand          | Harder to understand                    |
| Fewer relationships         | More relationships                      |
| Better for Power BI reports | More normalization increases complexity |
